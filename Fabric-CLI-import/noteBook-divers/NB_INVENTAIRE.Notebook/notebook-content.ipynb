```python
# Inventaire des tables silver et gold : noms, colonnes et une ligne d'exemple
for schema in ["silver", "gold"]:
    print(f"\n===== Schéma {schema} =====")
    for row in spark.sql(f"SHOW TABLES IN {schema}").collect():
        nom = f"{schema}.{row.tableName}"
        df = spark.table(nom)
        print(f"\n▶ {nom}")
        print("   colonnes :", ", ".join(f"{f.name} ({f.dataType.simpleString()})" for f in df.schema.fields))
        exemple = df.limit(1).collect()
        if exemple:
            print("   exemple  :", {k: str(v)[:40] for k, v in exemple[0].asDict().items()})
```
